```python
import pandas as pd
```

```python
df_laptops = pd.read_csv('laptop_price.csv', encoding='latin1')
```

```python
df_laptops.head(3)
```

# 1 isin(): Single Filtering

```python
# select Apple or HP laptops
df_laptops['Company'].isin(['Apple', 'HP'])
```

```python
# filter dataframe
df_laptops[df_laptops['Company'].isin(['Apple', 'HP'])]
```

# isin(): Multiple Filtering

```python
# find notebooks or ultabooks from Apple or HP
filter1 = df_laptops['Company'].isin(['Apple', 'HP'])
filter2 = df_laptops['TypeName'].isin(['Notebook', 'Ultrabook'])
```

```python
# filter dataframe
df_laptops[filter1 & filter2]
```

# 2 Find Duplicate Rows with the .duplicated() Method

## duplicated()

```python
# find duplicates in 1 column/series
df_laptops.duplicated('laptop_ID')
```

```python
# show elements in dataframe with duplicates in column 'laptop_ID'
df_laptops[df_laptops.duplicated('laptop_ID')]
```

```python
# duplicates in 2 or more columns
duplicated = df_laptops.duplicated(['Product', 'TypeName', 'Inches'])
```

```python
df_laptops[duplicated].sort_values(['Product', 'TypeName'])
```

# Examples 2 

```python
# Task: Find the cheapest and most expensive laptop of each company using the sort_values() and duplicated() methods (keep "first", "last" and False)
# sort dataframe asending by company and price
# cheapest first and most expensive last
df_laptops = df_laptops.sort_values(['Company', 'Price_euros'])
df_laptops
```

```python
# check all value categories in "Company" column
df_laptops.value_counts('Company')
```

```python
# duplicated values in "Company" columns (default -> keep='first')
duplicated_first = df_laptops.duplicated('Company')
```

```python
# show dataframe with duplicated values in "Company" column
df_laptops[duplicated_first]

# keep='first'(cheapest laptop per company)
# show dataframe with non duplicated values in "Company" column
# ~ means not
df_laptops[~duplicated_first]
df_laptops[~duplicated_first][['Company', 'Price_euros']]
```

```python
# check all value categories
# df_laptops[~duplicated].value_counts('Company')
```

```python
# keep='last' - keeps the last duplicate - which is reverse pf keep='first'
duplicated_last = df_laptops.duplicated('Company', keep='last')
# keep=False - delete all NO KEEPS
duplicated_false = df_laptops.duplicated('Company', keep=False)
```

```python
# keep='last'(most expensive laptop per company)
# show dataframe with non duplicated values in "Company" column
df_laptops[~duplicated_last]
df_laptops[~duplicated_last][['Company', 'Price_euros']]
```

```python
# keep=False
# show dataframe with non duplicated values in "Company" column
df_laptops[~duplicated_false]
```

```python
# all elements in company column are repeated at least once
df_laptops.value_counts('Company')
```

# 3 Drop Duplicate Elements with the .drop_duplicates() Method

```python
df_l = pd.read_csv('laptop_price.csv', encoding="latin1")
df_l.head(3)
```

## drop_duplicates()

```python
# drop duplicates in 2 or more columns
df_l.drop_duplicates(['Company'])
df_l.drop_duplicates(['Company'])[['Company', 'Price_euros']]
```

```python
# sort dataframe asending by company and price
# cheapest first and most expensive last
df_l = df_laptops.sort_values(['Company', 'Price_euros'])
```

```python
# cheapest: keep='first'
df_l.drop_duplicates(['Company'], keep='first')[['Company','Price_euros']]
```

```python
# most expensive keep='last'
df_l.drop_duplicates(['Company'], keep='last')[['Company', 'Price_euros']]
```

# Get and Count Unique Values The unique() and nunique() methods

```python
df_lt = pd.read_csv('laptop_price.csv', encoding="latin1")
```

# unique()
Return unique values of Series object. (Uniques are returned in order of appearance.)

```python
# get unique elements in the "Company" column
df_lt['Company'].unique()
```

```python
len(df_lt['Company'].unique())
```

# nunuque()
Return number of unique elements in the object (Excludes NA values by default)

```python
# Doing it the long way
# len(df_lt['Company'].unique())

# Doing it the short way
df_lt['Company'].nunique()

# get number of unique elements in the "Company" column
```
