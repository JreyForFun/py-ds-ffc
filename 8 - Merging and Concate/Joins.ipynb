```python
import pandas as pd
```

```python
df_movies = pd.read_csv('IMDb movies.csv', low_memory=False)
df_ratings = pd.read_csv('IMDb ratings.csv')
```

```python
# select columns
df_movies = df_movies[['imdb_title_id', 'title', 'year',
                       'genre', 'country']]

df_ratings = df_ratings[['imdb_title_id', 'total_votes', 'mean_vote']]
```

# merge()

```python
df1 = pd.DataFrame({'id': ['A', 'B', 'C', 'D'],
                    'age': [30, 23, 25, 22]})
df2 = pd.DataFrame({'id': ['C', 'D', 'E', 'F'],
                    'job': ['Doctor', 'Statistician',
                            'Accountant', 'Developer']})
```

```python
df1
```

```python
df2
```

# Inner Join

```python
df1.merge(df2, on='id', how='inner')
```

```python
# shape
print(df1.shape)
print(df2.shape)
print(df1.merge(df2, on='id', how='inner').shape)
```

```python
# merge df_movies and df_ratings (inner join)
df_movies.merge(df_ratings, on='imdb_title_id')
# pd.merge(df_movies, df_ratings, on='imdb_title_id')
```

# Outer join (full join)

```python
df1.merge(df2, on='id',  how='outer')
```

```python
# shape
print(df1.shape)
print(df2.shape)
print(df1.merge(df2, on='id', how='outer').shape)
```

```python
# 1. INNER JOIN (how='inner')
# Only keeps rows where the 'id' exists in BOTH tables (C and D)[cite: 2].
inner_df = pd.merge(df1, df2, on='id', how='inner')
print("--- Inner Join ---")
print(inner_df)


# 2. LEFT JOIN (how='left')
# Keeps EVERYTHING from the left table (df1: A, B, C, D)[cite: 2]. 
# Brings matching jobs over, and puts NaN where there is no match (A and B).
left_df = pd.merge(df1, df2, on='id', how='left')
print("\n--- Left Join ---")
print(left_df)


# 3. RIGHT JOIN (how='right')
# Keeps EVERYTHING from the right table (df2: C, D, E, F)[cite: 2]. 
# Brings matching ages over, and puts NaN where there is no match (E and F).
right_df = pd.merge(df1, df2, on='id', how='right')
print("\n--- Right Join ---")
print(right_df)


# 4. OUTER JOIN (how='outer')
# The "come one, come all" join. Keeps every single ID from both tables (A, B, C, D, E, F)[cite: 2].
# Fills in NaN anywhere data is missing on either side.
outer_df = pd.merge(df1, df2, on='id', how='outer')
print("\n--- Outer Join ---")
print(outer_df)
```

# Exlusive outer join(exclusive full join)


```python
df1.merge(df2, on='id', how='outer')
```

```python
# indicator=True

df1.merge(df2, on='id', how='outer', indicator=True)
```

```python
df1.merge(df2, on='id', how='outer', indicator=True).query("_merge=='left_only' or _merge=='right_only'")
```

```python
# shape
print(df1.shape)
print(df2.shape)
print(df1.merge(df2, on='id', how='outer',
          indicator=True).query("_merge=='left_only' or _merge=='right_only'").shape)
```

```python
# merge df_movies and df_ratings
df_movies.merge(df_ratings, on='imdb_title_id', how='outer', 
                indicator=True).query("_merge=='left_only' or _merge=='right_only'")
```
