```python
import pandas as pd
```

# Dataset Overview and Making Pivot Tables

```python
# read the csv file
df_popu = pd.read_csv('population_total.csv')
```

```python
# look at the dataset
df_popu.head()
```

```python
# Making a Pivot Table
# .pivot(): Returns reshaped DataFrame organized by given index / column values ("pivot without aggregation")

df_popu.isna().sum()
```

```python
df_popu['country'].nunique()
```

```python
df_popu.dropna(inplace=True)
```

```python
df_popu.isna().sum()
```

```python
# making a pivot table
df_pivot = df_popu.pivot(index='year', columns='country', values='population')
```

```python
# selecting some countries
df_pivot = df_pivot[['United States', 'India', 'China', 
                          'Indonesia', 'Brazil']]
```

```python
df_pivot
```

# Data Visualization with Pandas

# Lineplot

```python
df_pivot.plot(kind='line', xlabel='Years', ylabel='Population',
             title='Population (1955-2020', figsize=(8,4))
```

# Barplot
## Single Barplot

```python
# selecting only one year (2020)
df_pivot_2020 = df_pivot[df_pivot.index.isin([2020])]
```

```python
# showing new dataframe
df_pivot_2020
```

```python
# transpose dataframe from rows to columns or viceversa
df_pivot_2020 = df_pivot_2020.T
```

```python
df_pivot_2020.plot(kind='bar', color='orange',
                  xlabel='Years', ylabel='Population',
                  title='Population in 2020')
```

## Barplot grouped by “n” variables

```python
# selecting a few years
df_pivot_sample = df_pivot[df_pivot.index.isin([1980, 1990, 2000, 2010, 2020])]
```

```python
# making grouped barplot
df_pivot_sample = df_pivot[df_pivot.index.isin([1980, 1990, 2000, 2010, 2020])]
```

```python
# making grouped barplot
df_pivot_sample.plot(kind='bar', xlabel='Years', ylabel='Population',
                    figsize=(8,4))
```

# Piechart

```python
# changing column name
df_pivot_2020.rename(columns={2020:'2020'}, inplace=True)
```

```python
# making the plot
df_pivot_2020.plot(kind='pie', y='2020', 
                  title='Population in 2020 (%)')
```

# Boxplot
## Single Boxplot

```python
df_pivot['United States'].plot(kind='box', color='green',
                              ylabel='Population')
```

# Multiple boxplot

```python
df_pivot.plot(kind='box', xlabel='Countries',
              ylabel='Population')
```

# Histogram

```python
df_pivot[['United States', 'Indonesia']].plot(kind='hist',
                                              xlabel='Population')
```

# ScatterPlot

```python
# selecting a sample of the raw data
df_samp = df_popu[df_popu['country'].isin(['United States', 'India', 'China', 'Indonesia', 'Brazil'])]
```

```python
# creating the plot
df_samp.plot(kind='scatter',
              x='year',
              y='population',
              figsize=(10,8),
              color='green',
              s=80)
```
