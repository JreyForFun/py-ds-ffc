```python
import pandas as pd
import cufflinks as cf
import plotly.express as px
import plotly.io as pio
pio.templates.default = 'ggplot2'

cf.set_config_file(sharing='public', theme='ggplot', offline=True)
```

```python
cf.getThemes()
```

```python
# reading the csv file
df_population_raw = pd.read_csv('population_total.csv')
# dropping null values
df_population_raw.dropna(inplace=True)
# making a pivot table
df_pivot = df_population_raw.pivot(index='year', columns='country', values='population')

# selecting some countries
df_pivot = df_pivot[['United States', 'India', 'China',
                          'Indonesia', 'Brazil', 'Philippines']]

# showing pivot table
df_pivot
```

# Lineplot

```python
fig = px.line(df_pivot, title='Population 1955-2020 (%)', labels={'value': 'Population'}, template='ggplot2')
fig.show()
```

# Piechart

```python
fig = px.pie(
    df_pivot,
    
    template='ggplot2',
    title='Population Share'
)
fig.show()
```

# Barplot

```python
fig = px.bar(
    df_pivot,
    y="country",
    template='ggplot2',
    labels={'value': 'Population'}
)
fig.show()
```

# Histogram

```python
fig = px.histogram(
    df_pivot, 
    template='ggplot2',
    labels={'value': 'Population'}
)
fig.show()
```

# Scatter Plot

```python
fig = px.scatter(
    df_pivot,  
    template='ggplot2',
    labels={'value': 'Population'}
)
fig.show()
```

# BoxPlot

```python
fig = px.box(
    df_pivot, 
    template='ggplot2',
    labels={'value': 'Population'}
)
fig.show()
```
