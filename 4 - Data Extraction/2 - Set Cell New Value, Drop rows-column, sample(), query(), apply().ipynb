# 2 Drop Rows Or Columns from a Dataframe

```python
import pandas as pd
```

```python
# reads csv file
df = pd.read_csv('players_20.csv')
# set index
df.set_index('short_name', inplace=True)
# select column
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

```python
df.head()
```

# Drop Row

```python
# drop one row
# axis parameter # 1 = columns - 0 = rows
df.drop('L. Messi', axis=0) # old way

# index parameter
df.drop(index=['L. Messi']) # modern way
# safer
df.drop(['L. Messi']) # modert way too
```

```python
# Both of these do the exact same thing for multiple rows
# df.drop(['L. Messi', 'Cristiano Ronaldo'])
# df.drop(['L. Messi', 'Cristiano Ronaldo'], axis=0, inplace=True)
df.drop(index=['L. Messi', 'Cristiano Ronaldo']) # safer
```

```python
df
```

# Drop Columns

```python
# drop one column
# axis parameter
df.drop('long_name', axis=1)

# index parameter
df.drop(columns=['long_name'])
```

```python
# drop column by position/index (drop last column)
df.drop(df.columns[[-1]], axis=1)
```

```python

# drop two or more columns and update data (inplace=True)
df.drop(['long_name', 'dob'], axis=1, inplace=True)
```

```python
df
```

# Create Random Sample with the sample() Method

```python
import pandas as pd
```

```python
# reads csv file
df= pd.read_csv('players_20.csv')
# set index
df.set_index('short_name', inplace=True)
# select column
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

```python
# dataframe overview
df.head()
```

# sample()

```python
# extract 10 random elements from the "nationality" column
df['nationality'].sample(10, random_state=99)
```

```python
# extract a random 20% sample of the dataframe
df.sample(frac=0.2, random_state=99)
```

```python
# upsample: increase the sampling rate -> frac>1  (Note: replace parameter has to be True for frac parameter > 1)
df.sample(frac=2, replace=True, random_state=99)
```

# .Filter A DataFrame with the query() method

```python
# reads csv file
df = pd.read_csv('players_20.csv')
# set index
df.set_index('short_name', inplace=True)
# select column
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

```python
# show dataframe
df
```

# query()

```python
# select players older tha 34
df.query('age>34')
# exercise: write the equivalent boolean slicing
# df[df['age']>34]
```

```python
# select players older than 34 from Italy
df.query("age>34 and nationality=='Italy'")

# exercise: write the equivalent boolean slicing
# df[(df['age']>34) & (df['nationality']=='Italy')]
```

```python
#  add a not operator to the first example
df.query("not(age>34)")

# exercise: write the equivalent boolean slicing
# df[~(df['age']>34)]
```

```python
# convert height to meters and select those with height above 1.8
df.query("height_cm/100 > 1.8")

# exercise: write the equivalent boolean slicing
# df[df['height_cm']/100>1.8]
```

```python
# select players that were born after 1990

# check out data types
df.dtypes
```

```python
# conver "dob" column to datetime type
df['dob'] = df['dob'].astype('datetime64[ns]')
# alt and actualy good df['dob'] = pd.to_datetime(df['dob'])
# query
df.query('dob.dt.year > 1990')

# alt equivalent boolean slicing
df[df['dob'].dt.year>1990]
```
