# 2 Creating a Conditional Column from 2 Choices: np.where()

```python
import numpy as np
```

```python
# create an array based on price tiers
np.where(df_laptops['Price_euros'] > 2000, 'Expensive', 'Inexpensive')
```

```python
# # set it to a new column
df_laptops['Price_tier'] = np.where(df_laptops['Price_euros'] > 2000, 'Expensive', 'Inexpensive')
```

```python
df_laptops.head()
```

```python
df_laptops = df_laptops.drop(columns=['Price_tier'])
```

```python
df_laptops
```

```python
tiers = np.where(df_laptops['Price_euros'] > 2000, 'Expensive', 'Inexpensive')
df_laptops.insert(len(df_laptops.columns) - 1, 'Price_tier', tiers)
```

```python
df_laptops
```

# 3 Filter a DataFrame Baed on 2 or More Conditions

# Filter dataframe based on a condition

```python
# find Apple laptops
df_laptops['Company'] == 'Apple'
```

```python
# find laptops that cost more than 2000 euros
df_laptops['Price_euros'] > 2000
```

```python
df_laptops[df_laptops['Price_euros'] > 2000].sort_values('Price_euros')
```

# Filter dataframe based on multiple conditions

```python
# find Apple laptops that cost more than 2000 euros
df_laptops[(df_laptops['Company'] == 'Apple') & (df_laptops['Price_euros'] > 2000)]
```

# & (and) 

```python
# filer dataframe based on multiple conditions
df_laptops[(df_laptops['Company']=='Apple') & (df_laptops['Price_euros']>2000)]
```

# | (or)

```python
# find laptops Apple or Dell laptops
# filer dataframe based on multiple conditions
df_laptops['Company'] == 'Apple'
df_laptops['Company'] == 'Dell'

df_laptops[(df_laptops['Company'] == 'Apple') | (df_laptops['Company'] == 'Dell')]
```

```python
# find laptops from Apple or Dell that cost over 2000 euros
# find laptops from apple or dell
((df_laptops['Company'] == 'Apple') | (df_laptops['Company'] == 'Dell'))

# find that cost over 2000 euros
(df_laptops['Price_euros']>2000)

# combine
((df_laptops['Company'] == 'Apple') | (df_laptops['Company'] == 'Dell')) & (df_laptops['Price_euros']>2000)
```

```python
# filer dataframe based on multiple conditions
df_laptops[((df_laptops['Company'] == 'Apple') | (df_laptops['Company'] == 'Dell')) & (df_laptops['Price_euros']>2000)]
```

# 4 Creating a Conditional Column from More Than 2 Choices: np.select()

```python
# df_laptops = df_laptops.drop(columns=['Price_tier'], inplace=True)
```

```python
import numpy as np

# create an array based on multiple price tiers (+2 choices)

# create "conditions" and "values" variables
conditions = [
    (df_laptops['Price_euros']>3000),
    (df_laptops['Price_euros']>2000) & (df_laptops['Price_euros']<=3000),
    (df_laptops['Price_euros']>800) & (df_laptops['Price_euros']<=2000),
    (df_laptops['Price_euros']<=800),
]
choices = ['Too Expensive', 'Expensive', 'Affordable', 'Cheap']
```

```python
# set it to a new column
df_laptops['Price_tier'] = np.select(conditions, choices, default='Unknown')
```

```python
df_laptops
```

```python
# count values inside "price_tier" column
df_laptops.value_counts('Price_tier')
```

# Exercise

```python
# create an array based on multiple screen size tiers (+2 choices)
# Too Big>16, Big>14, Small>12, Too Small<12

# create "conditions" and "values" variables
conditions = [
    (df_laptops['Inches']>16),
    (df_laptops['Inches']>14) & (df_laptops['Inches']<=16),
    (df_laptops['Inches']>12) & (df_laptops['Inches']<=14),
    (df_laptops['Inches']<=12),
]

choices =['Too Big', 'Big', 'Small', 'Too Small']
```

```python
df_laptops['Screen_size'] = np.select(conditions, choices, default="Unknown")
```

```python
df_laptops
```

```python
df_laptops.value_counts('Screen_size')
```
