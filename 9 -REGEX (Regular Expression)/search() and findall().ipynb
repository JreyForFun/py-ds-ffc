# The re module

```python
import re
```

# re.search()

```python
text = """Hello World.
I love Python!!!!!
My first lucky number is 987-654-321
My second lucky number is 876-543-210
My third lucky number is 765-432-100
My fourth lucky number is 123-456-123-123-456
"""
```

```python
# return the first match
re.search(r'\d', text)
```

```python
# flags
re.search(r'^hello', text, flags=re.M)
```

# re.findall()

```python
# return all the matches
re.findall(r'\d', text)
#\d is any digit mean
```

```python
# return all the matches
re.findall(r'\d+', text)
```

```python
# Find every single individual letter
re.findall(r'[a-zA-Z]', text)
```

```python
# Find every word individual
# [a-zA-Z]+ means "match one or more letters in a row"
re.findall(r'[a-zA-Z]+', text)
```

```python
# \w+ grabs any block of letters, numbers, or underscores
print(re.findall(r'\w+', text))
```

```python
# Match only punctuation
text = """Hello World.
I love Python!!!!!
My first lucky number is 987-654-321
My second lucky number is 876-543-210
My third lucky number is 765-432-100
My fourth lucky number is 123-456-123-123-456
"""
re.findall(r'[^\w\s]', text)
```

```python
# Match right date format (MM-DD-YYYY)
text = '''
13-04-2021
2021-13-04
2021-04-13
'''
re.findall(r'\d{2}-\d{2}-\d{4}', text)
```

```python
# Match right username format (4-14 characters and only letters or digits)
text = '''
username10
abc
10
'''
re.findall(r'[a-zA-Z0-9]{4,14}', text)
```
