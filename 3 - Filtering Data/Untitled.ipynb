# 1 Filter dataframe based on a condition

```python
import pandas as pd
```

```python
df_laptops = pd.read_csv('laptop_price.csv', encoding='latin1')
```

```python
df_laptops.head(3)
```

# Filter dataframe based on a condition

```python
# find which rows have "Apple" in the "Company" column
df_laptops['Company']=='Apple'
```

```python
# filter dataframe based on condition
df_laptops[df_laptops['Company'] == 'Apple']
```

```python
# verify which companies are in the "Company" column with value_counts()
df_laptops[df_laptops['Company'] == 'Apple'].value_counts('Company')
```

# 2 Creating a Conditional Column from 2 Choices: np.where()

```python
import numpy as np
```

```python
# create an array based on price tiers
np.where(df_laptops['Price_euros'] > 2000, 'Expensive', 'Inexpensive')
```

```python
# # set it to a new column
df_laptops['Price_tier'] = np.where(df_laptops['Price_euros'] > 2000, 'Expensive', 'Inexpensive')
```

```python
df_laptops.head()
```

```python
df_laptops = df_laptops.drop(columns=['Price_tier'])
```

```python
df_laptops
```

```python
tiers = np.where(df_laptops['Price_euros'] > 2000, 'Expensive', 'Inexpensive')
df_laptops.insert(len(df_laptops.columns) - 1, 'Price_tier', tiers)
```

```python
df_laptops
```
